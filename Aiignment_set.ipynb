```python
# Create a set of first 10 natural numbers
natural_numbers = set(range(1, 11))
print(natural_numbers)
```

```python
# Add element 11 to set {1, 2, 3, 4, 5}
num={1, 2, 3, 4, 5}
num.add(11)
print(num)
```

```python
# Remove element 3 from set {1, 2, 3, 4, 5, 6}
num = {1, 2, 3, 4, 5, 6}
num.remove(3)
print(num)
```

```python
#  Find the intersection of {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
intersection = set1 & set2
print(intersection)
```

```python
# Find the difference between {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}
difference = set1 - set2
print(difference)   
```

```python
# Check if 5 is in set {1, 2, 3, 4, 6, 7, 8}
set3 = {1, 2, 3, 4, 6, 7, 8}
print(5 in set3)
```

```python
# Find the length of set {'a', 'b', 'c', 'd', 'e'}
set4 = {'a', 'b', 'c', 'd', 'e'}
print(len(set4))
```

```python
# Create a set of vowels from string "hello world"
vowels = set("hello world") & set("aeiou")
print(vowels)
```

```python
# Remove duplicates from list [1, 2, 2, 3, 4, 4, 5, 6, 6, 7] using set
list_with_duplicates = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
unique_set = set(list_with_duplicates)
print(unique_set)   
```

```python
# Check if {1, 2, 3} is a subset of {1, 2, 3, 4, 5, 6}
set5 = {1, 2, 3}
set6 = {1, 2, 3, 4, 5, 6}
print(set5.issubset(set6))
```
