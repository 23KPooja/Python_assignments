```python
# Create a dictionary of student names and their ages
student_ages = {
    "Alice": 20,
    "Bob": 22,
    "Charlie": 21
}   
print(student_ages)
```

```python
# Add a new key-value pair to dictionary {'a': 1, 'b': 2, 'c': 3}
student_ages["David"] = 23
print(student_ages)
```

```python
# Get all keys from dictionary {'name': 'John', 'age': 25, 'city': 'New York'}
x={'name': 'John', 'age': 25, 'city': 'New York'}
print(list(x.keys()))
```

```python
#  Get all values from dictionary {'python': 3, 'java': 2, 'c++': 1}
y = {'python': 3, 'java': 2, 'c++': 1}
print(list(y.values()))
```

```python
# Check if key 'age' exists in {'name': 'Alice', 'age': 30, 'city': 'London'}
x = {'name': 'Alice', 'age': 30, 'city': 'London'}
print('age' in x)
```

```python
#  Remove key 'temp' from {'a': 1, 'b': 2, 'temp': 3, 'c': 4}
z = {'a': 1, 'b': 2, 'temp': 3, 'c': 4}
del z['temp']
print(z)
```

```python
#  Find the sum of all values in {'math': 85, 'science': 92, 'english': 78}
z = {'math': 85, 'science': 92, 'english': 78}
print(sum(z.values()))
```

```python
#  Create a dictionary with squares of numbers 1 to 5
squares = {x: x**2 for x in range(1, 6)}
print(squares)
```

```python
# Count frequency of each character in string "hello"
text = "hello"
char_freq = {char: text.count(char) for char in set(text)}
print(char_freq)
```

```python
# Merge two dictionaries {'a': 1, 'b': 2} and {'c': 3, 'd': 4}
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```

```python
# Create a nested dictionary: {'person': {'name': 'Alice', 'age': 25}}
nested_dict = {'person': {'name': 'Alice', 'age': 25}}
print(nested_dict)
```
