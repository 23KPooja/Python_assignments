```python
principal = 1000
rate = 0.055
years = 3
compounding_per_year = 1  # 1 for annual, 12 for monthly

# Calculate final amount
amount = principal * (1 + rate / compounding_per_year) ** (compounding_per_year * years)
interest = amount - principal

print(f"Total Amount: ${amount:.2f}")    # Output: $1174.24
print(f"Earned Interest: ${interest:.2f}")  # Output: $174.24
```

```python
#  Find the hypotenuse of sides 3.5 and 4.2
n1=(3.5**2)
n2=(4.2**2)
hypotenuse = (n1+n2)**0.5
print(hypotenuse)  
```

```python
# Calculate the volume of a sphere with radius 7.8
radius = 7.8
volume = (4/3) * 3.14159 * (radius ** 3)
print(f"Volume of the sphere: {volume:.2f}")    
```
