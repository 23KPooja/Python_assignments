```python
# Write a program to check if a given year is a leap year or not.
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
    
```

```python
# Write a program that calculates BMI (Body Mass Index) and categorizes it.
height = float(input("Enter the person's height (in meters): "))
weight = float(input("Enter the person's weight (in kilograms): "))

bmi = weight / (height ** 2)

if bmi < 18.5:
    category = "Underweight"
elif 18.5 <= bmi < 25:
    category = "Normal weight"
elif 25 <= bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"The person's BMI is: {bmi:.2f}")
print(f"Category: {category}")
```

```python
# Write a program that prints the multiplication table for a given number (1 to 10).
number = int(input("Enter a number to print its multiplication table: "))
print(f"Multiplication table for {number}:")
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}") 

    
```

```python
# Write a program to calculate the factorial of a number using a for loop.
number = int(input("Enter a number to calculate its factorial: "))  
factorial = 1

for i in range(1, number + 1):
    factorial *= i

print(f"The factorial of {number} is: {factorial}")
```

```python
# Write a program that calculates the sum of all even numbers between 1 and n (inclusive), where n is provided by the user.
n = int(input("Enter a number: "))
sum_even = 0

for i in range(2, n + 1, 2):
    sum_even += i

print(f"The sum of all even numbers between 1 and {n} is: {sum_even}")
```
