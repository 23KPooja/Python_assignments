```python
# Write a program that calculates BMI (Body Mass Index) and categorizes it.
height = float(input("Enter the person's height (in meters): "))
weight = float(input("Enter the person's weight (in kilograms): "))

bmi = weight / (height ** 2)

if bmi < 18.5:
    category = "Underweight"
elif 18.5 <= bmi < 25:
    category = "Normal weight"
elif 25 <= bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"The person's BMI is: {bmi:.2f}")
print(f"Category: {category}")
```

```python
# Write a program to check if a given number is prime or not without using def.
# Hint: Check if n is divisible by any number from 2 to n-1. If no divisor found, it's prime. Optimize: check only up to √n.


n = int(input("Enter a number: "))
is_prime = True

if n < 2:
    is_prime = False
else:
    i = 2
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1

if is_prime:
    print(n, "is a prime number.")
else:
    print(n, "is not a prime number.")
```

```python
# Write a program that reverses the digits of a given number using a while loop(without using def).
n = int(input("Enter a number: "))
reversed_num = 0
negative = n < 0
n = abs(n)

while n > 0:
    digit = n % 10
    reversed_num = reversed_num * 10 + digit
    n //= 10

if negative:
    reversed_num = -reversed_num

print("Reversed number:", reversed_num)
```

```python
# Write a program to check if a given number is a palindrome or not.
n = int(input("Enter a number: "))
original = n
reversed_num = 0
n = abs(n)

while n > 0:
    digit = n % 10
    reversed_num = reversed_num * 10 + digit
    n //= 10

if original == reversed_num:
    print(f"{original} is a palindrome.")
else:
    print(f"{original} is not a palindrome.")
```

```python
# Write a program to check if a number is an Armstrong number (also called Narcissistic number).
n = int(input("Enter a number: "))
original = n
sum_of_powers = 0
n = abs(n)

# Find the number of digits
num_digits = len(str(n))

while n > 0:
    digit = n % 10
    sum_of_powers += digit ** num_digits
    n //= 10

if original == sum_of_powers:
    print(f"{original} is an Armstrong number.")
else:
    print(f"{original} is not an Armstrong number.")
```

```python
# Write a program to find the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.

# Algorithm:

# Divide the larger number by the smaller number
# Replace the larger number with the smaller number
# Replace the smaller number with the remainder
# Repeat until the remainder is 0
# The GCD is the last non-zero remainder
# Hint: While b != 0: store remainder (a%b), then a=b, b=remainder. When loop ends, a is the GCD.

a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print(f"The GCD is: {a}")
```

```python
# Write a program to check if a number is a perfect number.
num = int(input("Enter a number: "))
divisors_sum = 0

for i in range(1, num):
    if num % i == 0:
        divisors_sum += i

if divisors_sum == num:
    print(f"{num} is a perfect number.")
else:
    print(f"{num} is not a perfect number.")
```

```python
# Write a program to find the Least Common Multiple (LCM) of two numbers.
a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))

# Find the GCD first
while b != 0:
    remainder = a % b
    a = b
    b = remainder
gcd = a

# Calculate LCM using the formula: LCM(a, b) = (a * b) / GCD(a, b)
lcm = (a * b) // gcd

print(f"The LCM is: {lcm}")
```

```python
# Write a program that converts a number (0-999) to words.
num = int(input("Enter a number (0-999): "))    

# Define lists for number words
ones = ["", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine"]
teens = ["ten", "eleven", "twelve", "thirteen", "fourteen", "fifteen", "sixteen", "seventeen", "eighteen", "nineteen"]
tens = ["", "", "twenty", "thirty", "forty", "fifty", "sixty", "seventy", "eighty", "ninety"]

# Convert the number to words
if 0 <= num <= 999:
    if num == 0:
        print("zero")
    else:
        words = ""
        if num >= 100:
            words += ones[num // 100] + " hundred"
            num %= 100
        if num >= 20:
            words += " " + tens[num // 10]
            num %= 10
        if num >= 10:
            words += " " + teens[num - 10]
        else:
            words += " " + ones[num]
        print(words.strip())
else:
    print("Please enter a number between 0 and 999.")
```

```python
# Write a program to check if a number is a Strong number.
num = int(input("Enter a number: "))
sum_of_factorials = 0

temp = num
while temp > 0:
    digit = temp % 10
    factorial = 1
    for i in range(1, digit + 1):
        factorial *= i
    sum_of_factorials += factorial
    temp //= 10

if sum_of_factorials == num:
    print(f"{num} is a Strong number.")
else:
    print(f"{num} is not a Strong number.")
    
```

```python
# Write a program to print Pascal's Triangle up to n rows.
n = int(input("Enter the number of rows: "))

for i in range(n):
    # Print leading spaces
    for j in range(n - i - 1):
        print(" ", end="")
    
    # Print numbers for the current row
    num = 1
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    
    print()
```

```python
# Create a simple calculator program that:

# Shows a menu of operations: Add, Subtract, Multiply, Divide, Exit
# Takes two numbers as input
# Performs the selected operation
# Shows the result
# Repeats until the user chooses to exit
# Handles division by zero error
# Hint: Use while loop. Display menu, get choice. Use if-elif to perform operation. Check if divisor is 0 before division.

while True:
    print("\nSimple Calculator Menu:")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == '5':
        print("Exiting the calculator. Goodbye!")
        break

    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))

    if choice == '1':
        result = num1 + num2
        print(f"The result of addition is: {result}")
    elif choice == '2':
        result = num1 - num2
        print(f"The result of subtraction is: {result}")
    elif choice == '3':
        result = num1 * num2
        print(f"The result of multiplication is: {result}")
    elif choice == '4':
        if num2 == 0:
            print("Error: Division by zero is not allowed.")
        else:
            result = num1 / num2
            print(f"The result of division is: {result}")
    else:
        print("Invalid choice. Please select a valid option.")
```

```python
# Create an ATM machine simulator with the following features:

# Initial balance: 10000
# PIN: 1234 (3 attempts allowed)
# Menu options:
# Check Balance
# Withdraw Money (check for sufficient balance and valid amount)
# Deposit Money
# Change PIN
# Exit
# Keep transaction history (last 5 transactions)
# Input validation for all operations
# Hint: Use variables for balance, pin. Loop for PIN attempts. Use while loop for menu. Use list to store transaction history. Validate amounts > 0 and balance >= withdrawal.

import sys

balance = 10000
pin = 1234
attempts = 3
transaction_history = []

while attempts > 0:
    user_pin = int(input("Enter your PIN: "))
    if user_pin == pin:
        break
    else:
        attempts -= 1
        print(f"Invalid PIN. You have {attempts} attempts left.")

if attempts == 0:
    print("Too many failed attempts. Exiting.")
    sys.exit()

while True:
    print("\nATM Menu:")
    print("1. Check Balance")
    print("2. Withdraw Money")
    print("3. Deposit Money")
    print("4. Change PIN")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == '5':
        print("Exiting the ATM. Goodbye!")
        break

    if choice == '1':
        print(f"Your current balance is: ${balance:.2f}")

    elif choice == '2':
        amount = float(input("Enter the amount to withdraw: "))
        if amount <= 0:
            print("Invalid amount. Please enter a positive value.")
        elif amount > balance:
            print("Insufficient funds.")
        else:
            balance -= amount
            transaction_history.append(("Withdrawal", amount))
            if len(transaction_history) > 5:
                transaction_history.pop(0)
            print(f"Withdrawal successful. Your new balance is: ${balance:.2f}")

    elif choice == '3':
        amount = float(input("Enter the amount to deposit: "))
        if amount <= 0:
            print("Invalid amount. Please enter a positive value.")
        else:
            balance += amount
            transaction_history.append(("Deposit", amount))
            if len(transaction_history) > 5:
                transaction_history.pop(0)
            print(f"Deposit successful. Your new balance is: ${balance:.2f}")

    elif choice == '4':
        old_pin = int(input("Enter your current PIN: "))
        if old_pin == pin:
            new_pin = int(input("Enter your new PIN: "))
            pin = new_pin
            print("PIN changed successfully.")
        else:
            print("Incorrect current PIN.")

    else:
        print("Invalid choice. Please select a valid option.")
        
```
